```python
 #check if a number is a palindrome:  
number = int(input("enter number:"))
original  = number
reverse = 0
while number > 0:
    digit = number % 10
    number = number // 10
    reverse = reverse * 10 + digit
if original == reverse:
    print("palindrome")
else:
    print("not palindrome")
```
